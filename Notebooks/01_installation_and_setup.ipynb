# 01 · Installation & Setup

Welcome to the **TimesFM** example series. TimesFM is Google Research's
pretrained **foundation model for time-series forecasting**. It forecasts
*any* univariate series **zero-shot** — no training required.

This first notebook gets your environment ready and verifies your machine can
load the model.

**What you'll do here**
1. Install the package
2. Run the mandatory system preflight check
3. Confirm the imports work

> The TimesFM 2.5 weights (~800 MB) download from Hugging Face the first time
> you load the model and are cached in `~/.cache/huggingface/`. They are **not**
> stored in this repository.
